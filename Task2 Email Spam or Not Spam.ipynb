```python
import pandas as pd


```

```python
mail_data = pd.read_excel('Book1.xlsx').fillna('')

```

```python

mail_data = mail_data.rename(columns={mail_data.columns[0]: 'Email_Info'})

```

```python
mail_data['Sigmoid'] = mail_data['Email_Info'].str.extract(r'Sigmoid output=([0-9.]+)').astype(float)
```

```python
mail_data['Prediction'] = mail_data['Email_Info'].str.extract(r'Prediction: (\w+ ?\w*)')
```

```python
mail_data['Category'] = mail_data['Prediction'].map({'Spam': 0, 'Not Spam': 1})

```

```python
mail_data['Sigmoid (%)'] = (mail_data['Sigmoid'] * 100).round(1)
```

```python
display(mail_data[['Email_Info', 'Sigmoid', 'Sigmoid (%)', 'Prediction', 'Category']])
```

```python
import seaborn as sns
sns.set(style="whitegrid")
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(10,6))
sns.countplot(x='Prediction', data=mail_data, palette=['red','green'])
plt.title('Number of Emails: Spam vs Not Spam')
plt.xlabel('Prediction')
plt.ylabel('Count')
plt.show()
```

```python
plt.figure(figsize=(12,6))
sns.scatterplot(x=mail_data.index, y='Sigmoid', hue='Prediction', data=mail_data, palette=['red','green'])
plt.title('Email Spam Probability (Sigmoid Output)')
plt.xlabel('Email Index')
plt.ylabel('Sigmoid Output')
plt.legend(title='Prediction')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.histplot(data=mail_data, x='Sigmoid', hue='Prediction', bins=20, kde=True, palette=['red','green'])
plt.title('Distribution of Sigmoid Probabilities')
plt.xlabel('Sigmoid Output')
plt.ylabel('Number of Emails')
plt.show()
```
